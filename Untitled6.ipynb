```python
# =====================================
# 1. INSTALL LIBRARIES
# =====================================

# Removed "numpy<2" constraint to prevent binary incompatibilities with Gradio and Pandas on Python 3.13
!pip install -q "huggingface-hub==0.36.2" "transformers==4.57.3" sentence-transformers faiss-cpu pymupdf sentencepiece "gradio==5.44.1"


# =====================================
# 2. IMPORT LIBRARIES
# =====================================

import fitz
import gradio as gr
import faiss
import numpy as np
import torch

from sentence_transformers import SentenceTransformer
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM


# =====================================
# 3. LOAD AI MODELS
# =====================================

print("Loading embedding model...")

embedding_model = SentenceTransformer(
    "all-MiniLM-L6-v2"
)

print("Loading answer model...")

model_name = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_name)
answer_model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

answer_model.eval()

print("Models ready!")


# =====================================
# 4. PROCESS PDF FILES
# =====================================

def process_pdfs(files):
    if not files:
        return {}, "Please upload a PDF."

    chunks = []
    pages = []

    for file in files:
        doc = fitz.open(file)
        for page_number, page in enumerate(doc, start=1):
            text = page.get_text()
            for start in range(0, len(text), 800):
                chunk = text[start:start + 900]
                if chunk.strip():
                    chunks.append(chunk)
                    pages.append(page_number)
        doc.close()

    if not chunks:
        return {}, "No readable text found in these PDFs."

    print("Creating embeddings...")

    vectors = embedding_model.encode(
        chunks,
        normalize_embeddings=True,
        convert_to_numpy=True,
        show_progress_bar=True
    ).astype("float32")

    index = faiss.IndexFlatIP(vectors.shape[1])
    index.add(vectors)

    data = {
        "chunks": chunks,
        "pages": pages,
        "index": index
    }

    return data, f"Ready! Processed {len(chunks)} text chunks."


# =====================================
# 5. SEARCH PDF
# =====================================

def search_pdf(question, data):
    query = embedding_model.encode(
        [question],
        normalize_embeddings=True
    ).astype("float32")

    count = min(4, data["index"].ntotal)
    scores, ids = data["index"].search(query, count)

    results = []
    for score, idx in zip(scores[0], ids[0]):
        if idx != -1:
            results.append({
                "text": data["chunks"], [idx],
                "page": data["pages"][idx],
                "score": float(score)
            })

    return results

# Safety index fetch helper in search_pdf
def search_pdf(question, data):
    query = embedding_model.encode(
        [question],
        normalize_embeddings=True
    ).astype("float32")

    count = min(4, data["index"].ntotal)
    scores, ids = data["index"].search(query, count)

    results = []
    for score, idx in zip(scores[0], ids[0]):
        if idx != -1:
            results.append({
                "text": data["chunks"][idx],
                "page": data["pages"][idx],
                "score": float(score)
            })

    return results


# =====================================
# 6. GENERATE ANSWER
# =====================================

def ask_pdf(question, data):
    if not data:
        return "Please upload and process a PDF first."

    results = search_pdf(question, data)

    if not results or results[0]["score"] < 0.20:
        return "I couldn't find relevant information in the document."

    context = "\n".join(
        f"[Page {item['page']}] {item['text']}"
        for item in results
    )

    prompt = f"""
Answer using only the document context.
If the answer is missing, say you couldn't find it.

Context:
{context[:2500]}

Question: {question}

Answer:
"""

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    with torch.inference_mode():
        output = answer_model.generate(
            **inputs,
            max_new_tokens=150,
            num_beams=2
        )

    answer = tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )

    source_pages = sorted(set(
        item["page"] for item in results
    ))

    return (
        answer
        + "\n\nSource pages: "
        + ", ".join(map(str, source_pages))
    )


# =====================================
# 7. CHAT FUNCTION
# =====================================

def chat(message, history, data):
    if not message.strip():
        return "", history

    answer = ask_pdf(message, data)
    history = history + [
        {"role": "user", "content": message},
        {"role": "assistant", "content": answer}
    ]

    return "", history


# =====================================
# 8. BUILD USER INTERFACE
# =====================================

with gr.Blocks(title="AI PDF Assistant") as app:
    gr.Markdown("# 📚 AI PDF Assistant")
    gr.Markdown("Upload a PDF and ask questions about it.")

    document_state = gr.State({})

    with gr.Row():
        pdf_files = gr.File(
            label="Upload PDF",
            file_count="multiple",
            file_types=[".pdf"],
            type="filepath"
        )
        process_button = gr.Button("Process PDF", variant="primary")
        status = gr.Textbox(
            label="Status",
            interactive=False
        )

    chatbot = gr.Chatbot(
        label="Chat with your PDF",
        type="messages"
    )

    # Added a text box and button side by side inside a Row
    with gr.Row():
        question = gr.Textbox(
            placeholder="Ask something about your PDF...",
            scale=8
        )
        submit_button = gr.Button("Submit", scale=1, variant="primary")

    # Clicking button or pressing Enter does the exact same thing
    process_button.click(
        process_pdfs,
        inputs=[pdf_files],
        outputs=[document_state, status]
    )

    submit_event = submit_button.click(
        chat,
        inputs=[question, chatbot, document_state],
        outputs=[question, chatbot]
    )

    question.submit(
        chat,
        inputs=[question, chatbot, document_state],
        outputs=[question, chatbot]
    )


# =====================================
# 9. LAUNCH APP
# =====================================

app.launch(share=True)
```
